```python
import numpy as np
from collections import Counter

def calculate_impurity_metrics(labels):
    """
    Calculates both Gini Impurity and Entropy for a given list of class labels.
    """
    # Count the occurrences of each class
    total_items = len(labels)
    if total_items == 0:
        return 0.0, 0.0
        
    class_counts = Counter(labels)
    
    gini = 1.0
    entropy = 0.0
    
    print(f"--- Node Analysis ---")
    print(f"Total items: {total_items}")
    
    # Calculate probabilities and build the metrics
    for class_label, count in class_counts.items():
        probability = count / total_items
        print(f"Class '{class_label}': {count} items (Probability: {probability:.2f})")
        
        # Gini calculation: 1 - sum(p^2)
        gini -= probability ** 2
        
        # Entropy calculation: -sum(p * log2(p))
        if probability > 0:
            entropy -= probability * np.log2(probability)
            
    print("-" * 21)
    print(f"Gini Impurity: {gini:.4f} (Max 0.5000 for binary)")
    print(f"Entropy:       {entropy:.4f} (Max 1.0000 for binary)\n")
    
    return gini, entropy
```

```python

# --- Examples to Run ---

# Scenario 1: A perfectly pure node (Everyone paid)
print("Scenario 1: Perfectly Pure Node")
pure_node = ['Paid', 'Paid', 'Paid', 'Paid', 'Paid']
calculate_impurity_metrics(pure_node)
```

```python
# Scenario 2: A perfectly impure node (50/50 split)
print("Scenario 2: Maximum Impurity (50/50 Split)")
impure_node = ['Paid', 'Paid', 'Paid', 'Default', 'Default', 'Default']
calculate_impurity_metrics(impure_node)
```

```python
# Scenario 3: A partially pure node (Mostly paid, some defaults)
print("Scenario 3: Partially Pure Node (80/20 Split)")
partial_node = ['Paid'] * 8 + ['Default'] * 2
calculate_impurity_metrics(partial_node)
```

1. Impureza de Gini: O Risco de Errar o Chute (A Playlist no Celular)
Lembre-se que o Gini calcula a chance de você sortear um item, chutar o que ele é baseado na maioria, e errar.

Imagine que você está indo treinar na academia (que vi que é parte da sua rotina das 18:15 às 19:30). Você coloca o seu Spotify no modo aleatório.

Cenário de Baixo Gini (Puro): Você criou uma playlist chamada "Treino Pesado" que tem 95 músicas de Hip-Hop/Rap e apenas 5 de Samba. Se o Spotify sortear uma música aleatória e você "chutar" que é um Rap, a sua chance de errar é baixíssima. O Gini aqui é quase zero. A playlist é "pura".

Cenário de Alto Gini (Impuro): Você decide tocar a sua playlist "Curtidas" geral. Lá tem uma mistura caótica: 30% Rap, 30% Samba (pra praticar a dança), 20% MPB e 20% Rock. Se o Spotify sortear uma música e você chutar o gênero, a sua chance de errar é gigantesca. O Gini aqui é máximo (muito próximo de 1.0). A playlist está "impura".

Na vida real: O Gini mede o quão confiante você pode ser ao fazer uma aposta cega sobre algo que você tirou de um grupo misturado.

2. Entropia: O Número de Perguntas (Descobrindo o Jogo)
A Entropia mede quanta informação (ou quantas perguntas de sim/não) você precisa para acabar com a incerteza.

Imagine que é sexta-feira à noite, e a Ju (sua esposa) diz: "Comprei um jogo novo pro fim de semana!". Você não sabe qual é o jogo, e precisa descobrir fazendo o mínimo de perguntas de sim/não possível.

Cenário de Baixa Entropia (Poucas Perguntas): Você sabe que ela quase sempre compra jogos de tabuleiro e raramente compra jogos de videogame. A surpresa aqui é baixa. Você pergunta: "É de tabuleiro?". Se ela disser sim, a incerteza acabou em 1 bit de informação. A Entropia era baixa porque o comportamento dela é muito previsível.

Cenário de Alta Entropia (Muitas Perguntas): A Ju é totalmente imprevisível. Ela gosta de jogos de tabuleiro na mesma proporção que gosta de jogos de PC (tipo Civilization ou Palworld), e também gosta de jogos de Xbox Cloud. O grupo "Possíveis Jogos que a Ju Comprou" é muito misturado. Você vai precisar fazer várias perguntas: "É digital?" (Sim) -> "É de tiro/FPS tipo Valorant?" (Não) -> "É de estratégia?"... Você gastou muitos bits de informação (perguntas) para chegar à resposta. A Entropia era altíssima.

Na vida real: A Entropia mede a sua "surpresa". Se um evento é altamente previsível (o grupo é puro), a surpresa é zero e a Entropia é zero. Se tudo pode acontecer (o grupo é 50/50), a surpresa é máxima e a Entropia é alta.
